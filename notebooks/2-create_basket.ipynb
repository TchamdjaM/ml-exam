```python
#import library
from pathlib import Path
import pandas as pd
#import pickle
```

```python
#load data
order_products__prior=pd.read_csv("C:/Users/tcham/Desktop/ml-exam/data/order_products__prior.csv")
order_products__train=pd.read_csv("C:/Users/tcham/Desktop/ml-exam/data/order_products__train.csv")
```

```python
print("prior   :", order_products__prior.shape)
print("train   :", order_products__train.shape)
```

```python
#Filtrage having TOP N more frequents products in PRIOR)

TOP_N_PRODUCTS = 3000  # first try, more or less after result

prod_counts = order_products__prior["product_id"].value_counts()
kept_products = set(prod_counts.head(TOP_N_PRODUCTS).index)
```

```python
# putting the filter on  prior an train order

op_prior_f = order_products__prior[order_products__prior["product_id"].isin(kept_products)].copy()
op_train_f = order_products__train[order_products__train["product_id"].isin(kept_products)].copy()
```

```python
#have every order as list for the model
# list for the apriori and dictionary for the evaluation
# A- For evalset prior
baskets_prior_dict_ordered = (
    op_prior_f
    .sort_values(["order_id", "add_to_cart_order"])
    .groupby("order_id")["product_id"]
    .apply(list)
    .to_dict()
)
baskets_prior_list_ordered = list(baskets_prior_dict_ordered.values())

# B- For evalset train
baskets_train_dict_ordered = (
    op_train_f
    .sort_values(["order_id", "add_to_cart_order"])
    .groupby("order_id")["product_id"]
    .apply(list)
    .to_dict()
)
baskets_train_list_ordered = list(baskets_train_dict_ordered.values())
```

```python
## export my documents in csv and i place them in outputs
OUTPUT_DIR = Path(r"C:\Users\tcham\Desktop\ml-exam\outputs")

# 1) kept_products
pd.DataFrame({"kept_product_id": sorted(list(kept_products))}).to_csv(
    OUTPUT_DIR / "kept_products.csv", index=False
)

# 2) baskets_df : order_id + ordered list
baskets_prior_df = pd.DataFrame({
    "order_id": list(baskets_prior_dict_ordered.keys()),
    "product_id_list": list(baskets_prior_dict_ordered.values())
})

baskets_train_df = pd.DataFrame({
    "order_id": list(baskets_train_dict_ordered.keys()),
    "product_id_list": list(baskets_train_dict_ordered.values())
})

baskets_prior_df.to_csv(OUTPUT_DIR / "baskets_prior_ordered.csv", index=False)
baskets_train_df.to_csv(OUTPUT_DIR / "baskets_train_ordered.csv", index=False)

# 3) Export "dict" (baskets_*_df with clear names)
pd.DataFrame({
    "order_id": list(baskets_prior_dict_ordered.keys()),
    "product_id_list": list(baskets_prior_dict_ordered.values())
}).to_csv(OUTPUT_DIR / "baskets_prior_dict_ordered.csv", index=False)

pd.DataFrame({
    "order_id": list(baskets_train_dict_ordered.keys()),
    "product_id_list": list(baskets_train_dict_ordered.values())
}).to_csv(OUTPUT_DIR / "baskets_train_dict_ordered.csv", index=False)

# 4) Export "list" : 1 ligne = 1 order
pd.DataFrame({"transaction": baskets_prior_list_ordered}).to_csv(
    OUTPUT_DIR / "baskets_prior_list_ordered.csv", index=False
)

pd.DataFrame({"transaction": baskets_train_list_ordered}).to_csv(
    OUTPUT_DIR / "baskets_train_list_ordered.csv", index=False
)
```
