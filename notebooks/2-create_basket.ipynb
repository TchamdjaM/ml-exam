```python
#import library
from pathlib import Path
import pandas as pd
```

```python
#load data csv that will be use to create basket
orders=pd.read_csv("../data/orders.csv")
products=pd.read_csv("../data/products.csv")
order_products__prior=pd.read_csv("../data/order_products__prior.csv")
order_products__train=pd.read_csv("../data/order_products__train.csv")
```

```python
#Filtrage to having TOP N more frequents products in prior to avoid product rarely command. Choosing 3000 here

TOP_N_PRODUCTS = 3000  # first try, more or less after result

# putting the filter on  prior an train order for no error in the evaluation

top_products = order_products__prior["product_id"].value_counts().head(TOP_N_PRODUCTS).index

prior_small = order_products__prior[order_products__prior["product_id"].isin(top_products)].copy()
train_small = order_products__train[order_products__train["product_id"].isin(top_products)].copy()

print("prior_small:", prior_small.shape)
print("train_small:", train_small.shape)
```

```python
#creation of basket by transform order_id in list of product_id
def build_baskets(df):
    baskets = (
        df.groupby("order_id")["product_id"]
        .apply(lambda s: sorted(set(s.tolist())))
        .reset_index()
        .rename(columns={"product_id": "basket"})
    )
    return baskets

prior_baskets = build_baskets(prior_small)
train_baskets = build_baskets(train_small)

display(prior_baskets.head())
display(train_baskets.head())

print("nb prior baskets:", len(prior_baskets))
print("nb train baskets:", len(train_baskets))
```

```python
#Save the tables create in csv for the model utilisation in a outputs folder

prior_baskets.to_csv("../outputs/prior_baskets.csv", index=False)
train_baskets.to_csv("../outputs/train_baskets.csv", index=False)

products_small = products[products["product_id"].isin(top_products)].copy()
products_small.to_csv("../outputs/products_small.csv", index=False)

print("Saved to: ../outputs")
```
